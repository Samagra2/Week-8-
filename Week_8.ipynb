Multi-Dataset Business Analysis Capstone

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("default")
```

```python
house_df = pd.read_csv("/content/house_prices.csv")
churn_df = pd.read_csv("/content/customer_churn.csv")
sales_df = pd.read_csv("/content/sales_data.csv")

print("House Prices:", house_df.shape)
print("Customer Churn:", churn_df.shape)
print("Sales Data:", sales_df.shape)
```

```python
house_df = house_df.dropna()
churn_df = churn_df.dropna()
sales_df = sales_df.dropna()

house_df.to_csv("cleaned_house_prices.csv", index=False)
churn_df.to_csv("cleaned_customer_churn.csv", index=False)
sales_df.to_csv("cleaned_sales_data.csv", index=False)
```

```python
print("""
BUSINESS PROBLEMS:
1. What factors influence house prices?
2. Why are customers churning?
3. What drives sales performance?
""")
```

```python
plt.figure(figsize=(6,4))
sns.histplot(house_df['Price'], bins=20)
plt.title("House Price Distribution")
plt.show()

plt.figure(figsize=(6,4))
sns.scatterplot(x='Area', y='Price', data=house_df)
plt.title("Area vs Price")
plt.show()
```

```python
plt.figure(figsize=(6,4))
churn_df['Churn'].value_counts().plot(kind='bar')
plt.title("Customer Churn Distribution")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(x='Churn', y='MonthlyCharges', data=churn_df)
plt.title("Monthly Charges vs Churn")
plt.show()
```

```python
print(sales_df.columns)
sales_df.head()
```

```python
sales_df.columns = sales_df.columns.str.strip().str.lower()
print(sales_df.columns)
```

```python
plt.figure(figsize=(6,4))
sns.barplot(
    x='region',
    y='total_sales',
    data=sales_df,
    estimator=sum
)
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

```python
sales_df['date'] = pd.to_datetime(sales_df['date'])

daily_sales = sales_df.groupby('date')['total_sales'].sum().reset_index()

plt.figure(figsize=(6,4))
sns.lineplot(
    x='date',
    y='total_sales',
    data=daily_sales
)
plt.title("Sales Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.show()
```

```python
# 1. Correlation Analysis (House Prices)
plt.figure(figsize=(6,4))
numeric_house_df = house_df.select_dtypes(include=['int64', 'float64'])
sns.heatmap(numeric_house_df.corr(), annot=True, cmap='coolwarm')
plt.title("House Price Correlation")
plt.show()

# 2. Churn Rate
churn_rate = churn_df['Churn'].value_counts(normalize=True) * 100
print("Churn Rate (%)")
print(churn_rate)

# 3. Sales Aggregation
region_sales = sales_df.groupby('region')['total_sales'].sum()
print("Total Sales by Region")
print(region_sales)
```

```python
print("""
KEY INSIGHTS:
- Larger area significantly increases house prices
- Month-to-month customers churn more
- High monthly charges increase churn risk
- Certain regions consistently outperform in sales
""")
```

```python
print("""
RECOMMENDATIONS:
1. Real Estate: Focus marketing on high-area properties
2. Churn: Offer discounts to high-risk customers
3. Sales: Replicate strategies from top-performing regions

IMPLEMENTATION PLAN:
- Phase 1: Identify high-risk segments
- Phase 2: Apply targeted offers
- Phase 3: Monitor KPIs monthly
""")
```
